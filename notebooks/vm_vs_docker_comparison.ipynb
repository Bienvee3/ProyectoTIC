```python
import pandas as pd
import matplotlib.pyplot as plt
import statistics
import sys
print(sys.executable)

def calcularMedia(columna):
    return statistics.mean(columna)

def main():
    # Cargar datos (CORREGIDO: agregada la barra diagonal faltante)
    vm = pd.read_csv("../results/benchmark_snake_vm.csv")
    docker = pd.read_csv("../results/benchmark_snake_docker.csv")

    # Calcular medias
    vm_cpu_media = calcularMedia(vm["cpu_percent"])
    docker_cpu_media = calcularMedia(docker["cpu_percent"])

    vm_ram_media = calcularMedia(vm["memory_percent"])
    docker_ram_media = calcularMedia(docker["memory_percent"])

    vm_latency_media = calcularMedia(vm["latency"])
    docker_latency_media = calcularMedia(docker["latency"])

    # Gráfico 1: Uso de CPU
    plt.figure(figsize=(10, 4))
    plt.plot(vm["cpu_percent"], label="VM CPU")
    plt.plot(docker["cpu_percent"], label="Docker CPU")
    plt.axhline(vm_cpu_media, color='blue', linestyle='--', label=f"Media VM CPU ({vm_cpu_media:.2f}%)")
    plt.axhline(docker_cpu_media, color='orange', linestyle='--', label=f"Media Docker CPU ({docker_cpu_media:.2f}%)")
    plt.legend()
    plt.title("Uso de CPU: VM vs Docker")
    plt.xlabel("Muestras")
    plt.ylabel("% CPU")
    plt.grid(True)
    plt.show()

    # Gráfico 2: Uso de RAM
    plt.figure(figsize=(10, 4))
    plt.plot(vm["memory_percent"], label="VM RAM")
    plt.plot(docker["memory_percent"], label="Docker RAM")
    plt.axhline(vm_ram_media, color='blue', linestyle='--', label=f"Media VM RAM ({vm_ram_media:.2f}%)")
    plt.axhline(docker_ram_media, color='orange', linestyle='--', label=f"Media Docker RAM ({docker_ram_media:.2f}%)")
    plt.legend()
    plt.title("Uso de RAM: VM vs Docker")
    plt.xlabel("Muestras")
    plt.ylabel("% RAM")
    plt.grid(True)
    plt.show()

    # Gráfico 3: Latencia
    plt.figure(figsize=(10, 4))
    plt.plot(vm["latency"], label="VM Latencia")
    plt.plot(docker["latency"], label="Docker Latencia")
    plt.axhline(vm_latency_media, color='blue', linestyle='--', label=f"Media VM Latencia ({vm_latency_media:.2f} ms)")
    plt.axhline(docker_latency_media, color='orange', linestyle='--', label=f"Media Docker Latencia ({docker_latency_media:.2f} ms)")
    plt.legend()
    plt.title("Latencia: VM vs Docker")
    plt.xlabel("Muestras")
    plt.ylabel("ms")
    plt.grid(True)
    plt.show()

main()
```
